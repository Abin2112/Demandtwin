# Synthetic adaptive experiments
Compare policies under the same synthetic oracle and response budget.

```python
from pathlib import Path
import json
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "PROJECT_STATUS.md").exists())
results = root / "research" / "results"
import pandas as pd
record = json.loads((results / "choice_experiments.json").read_text())
assert record["synthetic"]
series = {policy: pd.DataFrame(points).set_index("observations")["share_mae"]
          for policy, points in record["experiment_2"]["policies"].items()}
pd.DataFrame(series).plot(ylabel="Synthetic share MAE")
```
